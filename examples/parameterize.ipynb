# Ligand preparation in Jupyter

Select a Python kernel with ligtools, Psi4, AmberTools and Open Babel installed.
Run the cells in order. No BLAS environment variables or multiprocessing setup
are needed. The initial pool is reduced here for a quick workflow check.

```python
from ligtools import Smi2Gmx
```

```python
ligands = Smi2Gmx(
    smiles=['CC(=O)C(=O)[O-]', 'CC(=O)[PH](=O)[O-]',
            'CC(=O)P(OC)(=O)[O-]', 'CC(=O)P(C)(=O)[O-]'],
    res_names=['PYR', 'API', 'MPO', 'MPI'],
    work_dir='./output_param/',
    n_conformer_pool=20,
    n_conformer_threads=4,
    max_n_cores=2,
)
ligands.draw_smiles()
```

```python
ligands.run_psi4()
```

```python
ligands.run_resp()
```

```python
ligands.run_acpype(parallel=True)
```

```python
ligands.get_itps()
ligands.get_gros()
ligands.get_pdbs()
ligands.get_rtps()
ligands.draw_smiles()
```
